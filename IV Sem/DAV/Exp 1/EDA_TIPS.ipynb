# Exploratory Data Analysis (EDA) on TIPS Dataset

This notebook performs EDA on the TIPS dataset as per academic requirements.

## Import Libraries

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Dataset

```
df = pd.read_csv('tips.csv')
df.head()
```

## Univariate Non-Graphical Analysis

```
df['total_bill'].describe()
```

```
df['day'].value_counts()
```

## Univariate Graphical Analysis

```
plt.hist(df['total_bill'], bins=10)
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

```
plt.boxplot(df['tip'])
plt.ylabel('Tip')
plt.show()
```

## Multivariate Non-Graphical Analysis

```
df[['total_bill','tip','size']].corr()
```

```
df.groupby('day')['tip'].mean()
```

## Multivariate Graphical Analysis

```
plt.scatter(df['total_bill'], df['tip'])
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```
sns.boxplot(x='day', y='total_bill', data=df)
plt.show()
```

## Conclusion
EDA reveals strong relationships between total bill and tip, supporting descriptive analytics.
